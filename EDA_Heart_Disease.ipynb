```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Đọc dữ liệu gốc 
df_raw = pd.read_csv('data/heart.csv')
print("Kích thước ban đầu:", df_raw.shape)

# 2. Chứng minh phát hiện dòng trùng lặp 
duplicates = df_raw.duplicated().sum()
print(f"Phát hiện {duplicates} dòng bị nhân bản sai lệch!")

# 3. Làm sạch dữ liệu
df_clean = df_raw.drop_duplicates().reset_index(drop=True)
df_clean = df_clean[(df_clean['thal'] != 0) & (df_clean['ca'] != 4)]

# 4. Xuất bảng thống kê mô tả 
display(df_clean.describe().T)

# 5. Vẽ Ma trận tương quan (Correlation Heatmap)
plt.figure(figsize=(12, 8))
sns.heatmap(df_clean.corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Ma Trận Tương Quan Giữa Các Chỉ Số Y Tế')
plt.show()
```
